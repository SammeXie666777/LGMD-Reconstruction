```python
## View morphology with h.PlotShape
# Create arbi neuron
class Cell:
    def __int__(self): # initiate
        main = h.Section (name = 'main',cell = self)
        dend1 = h.Section (name = 'dend1',cell = self)
        dend2 = h.Section (name = 'dend2',cell = self)
        
        dend1.connect (main)
        dend2.connect (main)
        
        main.diam = 10 * um
        dend1.diam = 2 * um
        dend2.diam = 2 * um
        # store section
        self.main = main;
        self.dend1 = dend1;
        self.dend2 = dend2;
        self.all = main.wholetree()

my_cell = Cell()
ps = h.PlotShape (False)
ps.plot(plotly).show()

## View Voltage, sodium
my_cell.main.v = 50
my_cell.dend1.v = 0
my_cell.dend2.v = -65

ps = h.PlotShape (False)
ps.variale('v')
ps.scale(-75,75)
ps.plot(plotly).show()
```
